```python
import os
import pandas as pd
from DeepPurpose import utils
from DeepPurpose import DTI as models
import numpy as np
import time
```

```python
#Make the function to get dti virtual screening
common_path = '/home/rmall/TII/Projects/Raghav/Druggability/'

def DTI_pred(drugs, targets, model_name, drug_names, target_names):
    model = models.model_pretrained(common_path+"Models/pretrained_models/"+model_name)
    model.config['batch_size']=256
    print(model.config)

    #Get predictions using virtual screening
    y_pred = models.virtual_screening(drugs, targets, model, drug_names = drug_names, target_names = target_names, result_folder = common_path+"Results/Raj_Analysis/"+model_name, convert_y = False, output_num_max = 10, verbose = True)
    return(y_pred)
```

```python
#Load the drug SMILES dataset and protein sequence dataset
drug_target_interactions_df = pd.read_csv(common_path+"/Data/drug_protein_interactions.csv",header="infer",sep="|")
interest_drugs_df = pd.read_csv(common_path+"/Results/Raj_Analysis/interest_drugs.csv",header="infer",sep=",")

#Get list of drug smiles and drug ids
drug_smiles = drug_target_interactions_df["smiles"].tolist()
drug_names = drug_target_interactions_df["name"].tolist()

interest_drug_smiles = interest_drugs_df["SMILES"].tolist()
interest_drug_names = interest_drugs_df["Name"].tolist()
print(interest_drug_smiles)
print(interest_drug_names)

drug_smiles = drug_smiles + interest_drug_smiles
drug_names = drug_names + interest_drug_names

N_drugs = len(drug_smiles)
print(N_drugs)

#Print the no of drugs
print(drug_smiles[::-1])
print(drug_names[::-1])

#filtered_alphafold_df = pd.read_csv("../Data/Filtered_AlphaFold_UniProtKB_with_Sequence.csv",header="infer",sep=",")
#filtered_alphafold_df.head()

#duplicate_sequence_df = filtered_alphafold_df.loc[filtered_alphafold_df.groupby('Protein_Sequence').UniProtId.transform('nunique').gt(1)]
#duplicate_sequence_df.to_csv("../Data/Filtered_AlphaFold_UniProtKB_with_Duplicate_Sequences.csv",index=False,sep=",")

#rev_filtered_alphafold_df = filtered_alphafold_df.drop_duplicates(subset=['Protein_Sequence'])
#print(rev_filtered_alphafold_df.shape)

interest_proteins_df = pd.read_csv(common_path+"/Results/Raj_Analysis/interest_proteins.csv",header="infer",sep=",")

N_proteins = interest_proteins_df.shape[0]
protein_seqs = interest_proteins_df["ProteinSequence"].tolist()
protein_names = interest_proteins_df["Protein"].tolist()

revised_protein_seqs = protein_seqs * N_drugs
revised_protein_names = protein_names * N_drugs
print(len(revised_protein_seqs))
```

```python
#Make the large list for smiles and drug names
revised_drug_smiles, revised_drug_names=[],[]
for i in range(N_drugs):
    for j in range(N_proteins):
        revised_drug_smiles.append(drug_smiles[i])
        revised_drug_names.append(drug_names[i])

#Where is LOC14 located in drug names
print(drug_names.index("ZAPi"))
```

```python
model_names = ["CNN_CNN_BindingDB_IC50","MPNN_CNN_DAVIS","Morgan_CNN_BindingDB_IC50","Morgan_AAC_BindingDB_IC50","Daylight_AAC_BindingDB_IC50","CNN_CNN_DAVIS",
               "CNN_CNN_BindingDB","Morgan_CNN_BindingDB","Morgan_CNN_DAVIS","MPNN_CNN_BindingDB","MPNN_CNN_KIBA","Transformer_CNN_BindingDB",
               "Daylight_AAC_DAVIS","Daylight_AAC_KIBA","Daylight_AAC_BindingDB","Morgan_AAC_BindingDB","Morgan_AAC_KIBA","Morgan_AAC_DAVIS"]

#Already ran code for CNN_CNN_BindingDB
for i in range(0,1):
    model_name = model_names[i]
    start_time = time.time()
    predictions = DTI_pred(revised_drug_smiles, revised_protein_seqs, model_name, revised_drug_names, revised_protein_names)
    finish_time = time.time()-start_time

    #Move the results accordingly
    os.system("mv /home/rmall/TII/Projects/Raghav/Druggability/Results/Raj_Analysis/"+model_name+"/virtual_screening.txt /home/rmall/TII/Projects/Raghav/Druggability/Results/Raj_Analysis/"+model_name+"/"+model_name+"_virtual_screening.csv")
    
```

```python
import gc

model_names = ["CNN_CNN_BindingDB_IC50","MPNN_CNN_DAVIS","Morgan_CNN_BindingDB_IC50","Morgan_AAC_BindingDB_IC50","Daylight_AAC_BindingDB_IC50","CNN_CNN_DAVIS",
               "CNN_CNN_BindingDB","Morgan_CNN_BindingDB","Morgan_CNN_DAVIS","MPNN_CNN_BindingDB","MPNN_CNN_KIBA","Transformer_CNN_BindingDB",
               "Daylight_AAC_DAVIS","Daylight_AAC_KIBA","Daylight_AAC_BindingDB","Morgan_AAC_BindingDB","Morgan_AAC_KIBA","Morgan_AAC_DAVIS"]

zbp1_df = pd.DataFrame()
nlrp3_df = pd.DataFrame()

for model_name in model_names:
    #Performing analysis for which model
    print("Currently working on model: "+model_name)
    #Read the big data
    big_df = pd.read_csv(common_path+"Results/Raj_Analysis/"+model_name+"/"+model_name+"_virtual_screening.csv", header="infer",sep="|")
    
    #Remove auxillary columns
    big_df = big_df.drop(columns = ["Unnamed: 0","Unnamed: 5"],axis=1)
    big_df.columns = ["Rank","Drug Name","Target Name","Binding Score"]

    #Remove the rankings
    big_df = big_df.drop(columns = ["Rank"],axis=1)

    #Remove white spaces from drug name
    big_df["Drug Name"] = big_df["Drug Name"].str.replace(r'\s+', '', regex=True)

    #Remove white spaces from target name
    big_df["Target Name"]=big_df["Target Name"].str.replace(r'\s+', '', regex=True)

    #Remove all NAs
    big_df = big_df.dropna(how="all")
    big_df["Binding Score"]=big_df["Binding Score"].astype(float)

    ##If after removing NAs still NAs remain as string, then replace with 0
    #vals = np.where(big_df["Binding Score"].isnull())
    #if (len(vals)>0):
    #    big_df.loc[vals[0],2]=0   #3rd column is Binding Score
    
    #Convert dataframe to matrix
    pivoted_big_df = big_df.pivot(index="Drug Name",columns="Target Name",values="Binding Score")
    del big_df
    gc.collect()
    
    #Create a dataframe which contains drug names and columns are model names
    zbp1_df = pd.concat([zbp1_df,pivoted_big_df["ZBP1"]],axis=1)
    
    #Find the rank of LOC14 in pivoted_big_df
    pivoted_big_df = pivoted_big_df.sort_values(by="ZBP1",ascending=False)
    print(pivoted_big_df.index.tolist().index("ZAPi"))

    nlrp3_df = pd.concat([nlrp3_df,pivoted_big_df["NLRP3"]],axis=1)
    pivoted_big_df = pivoted_big_df.sort_values(by="NLRP3",ascending=False)
    print(pivoted_big_df.index.tolist().index("LOC14"))

    del pivoted_big_df
    gc.collect()
```

```python
zbp1_df.columns = model_names
nlrp3_df.columns = model_names

zbp1_df.to_csv(common_path+"Results/Raj_Analysis/ZBP1_Drug_Predictions.csv",index=True,sep=",")
nlrp3_df.to_csv(common_path+"Results/Raj_Analysis/NLRP3_Drug_Predictions.csv",index=True,sep=",")

zbp1_df[["Morgan_AAC_BindingDB_IC50","Morgan_CNN_BindingDB_IC50"]].to_csv(common_path+"Results/Raj_Analysis/ZBP1_Drug_Predictions_Subset.csv",index=True,sep=",")
nlrp3_df[["MPNN_CNN_BindingDB","Morgan_AAC_KIBA"]].to_csv(common_path+"Results/Raj_Analysis/NLRP3_Drug_Predictions_Subset.csv",index=True,sep=",")
```
